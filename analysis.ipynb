# Feature Engineering

Feature engineering involves creating additional variables from the existing dataset to support deeper business analysis.

The following features are created:

- **Shipping_Lead_Time**
- **Factory_Region_Route**
- **Factory_State_Route**

These features improve the analysis of shipping efficiency and regional logistics performance.

### Code Implementation

```python
df["Order Date"] = pd.to_datetime(df["Order Date"], dayfirst=True)
df["Ship Date"] = pd.to_datetime(df["Ship Date"], dayfirst=True)

df["Shipping_Lead_Time"] = (
    df["Ship Date"] - df["Order Date"]
).dt.days

df["Factory_Region_Route"] = (
    df["Division"] + " → " + df["Region"]
)

df["Factory_State_Route"] = (
    df["Division"] + " → " + df["State/Province"]
)

print(df.head())
```

# Key Performance Indicators (KPIs)

This section summarizes the overall business performance using key metrics such as total shipments, sales, profit, units sold, and average shipping lead time.

```python
# ==========================
# KPI CALCULATIONS
# ==========================

total_shipments = len(df)
total_sales = df["Sales"].sum()
total_profit = df["Gross Profit"].sum()
total_units = df["Units"].sum()

average_sales = df["Sales"].mean()
average_lead_time = df["Shipping_Lead_Time"].mean()

print("="*50)
print("KEY PERFORMANCE INDICATORS")
print("="*50)

print(f"Total Shipments    : {total_shipments:,}")
print(f"Total Sales        : ${total_sales:,.2f}")
print(f"Total Profit       : ${total_profit:,.2f}")
print(f"Total Units        : {total_units:,}")
print(f"Average Sales      : ${average_sales:.2f}")
print(f"Average Lead Time  : {average_lead_time:.2f} days")
```

# Region Analysis

This section compares shipment volume, sales, profit, and average shipping lead time across different regions.

```python
region_summary = (
    df.groupby("Region")
      .agg(
          Total_Shipments=("Order ID","count"),
          Average_Lead_Time=("Shipping_Lead_Time","mean"),
          Total_Sales=("Sales","sum"),
          Total_Units=("Units","sum"),
          Total_Profit=("Gross Profit","sum")
      )
      .sort_values(by="Total_Sales", ascending=False)
)

display(region_summary)
```

```python
plt.figure(figsize=(8,5))

region_summary["Total_Sales"].plot(kind="bar")

plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Sales")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

# Ship Mode Analysis

This section evaluates the performance of each shipping method based on shipments, sales, profit, and average lead time.

```python
ship_mode_summary = (
    df.groupby("Ship Mode")
      .agg(
          Total_Shipments=("Order ID","count"),
          Average_Lead_Time=("Shipping_Lead_Time","mean"),
          Total_Sales=("Sales","sum"),
          Total_Profit=("Gross Profit","sum")
      )
      .sort_values(by="Total_Sales", ascending=False)
)

display(ship_mode_summary)
```

```python
plt.figure(figsize=(8,5))

ship_mode_summary["Total_Sales"].plot(kind="bar")

plt.title("Total Sales by Ship Mode")
plt.xlabel("Ship Mode")
plt.ylabel("Sales")

plt.xticks(rotation=20)

plt.tight_layout()
plt.show()
```

# Monthly Sales Trend

Monthly sales are analyzed to identify trends and seasonal business performance over time.

```python
df["Month"] = df["Order Date"].dt.to_period("M").astype(str)

monthly_sales = (
    df.groupby("Month")["Sales"]
      .sum()
      .reset_index()
)

display(monthly_sales)
```

```python
plt.figure(figsize=(12,5))

plt.plot(
    monthly_sales["Month"],
    monthly_sales["Sales"],
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

# Product Analysis

This section identifies the top-performing products based on total sales, units sold, and profit.

```python
top_products = (
    df.groupby("Product Name")
      .agg(
          Total_Sales=("Sales","sum"),
          Total_Units=("Units","sum"),
          Total_Profit=("Gross Profit","sum")
      )
      .sort_values(by="Total_Sales", ascending=False)
      .head(10)
)

display(top_products)

```

```python
plt.figure(figsize=(12,6))

top_products["Total_Sales"].plot(kind="bar")

plt.title("Top 10 Products by Sales")
plt.xlabel("Product")
plt.ylabel("Sales")

plt.xticks(rotation=75)

plt.tight_layout()

plt.show()
```
